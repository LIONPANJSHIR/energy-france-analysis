```python
from pathlib import Path

import pandas as pd


file_path = Path(
    "../data/raw/eCO2mix_RTE_Annuel-Definitif_2024.xls"
)

# Lecture de la ligne d'en-tête pour récupérer les 40 noms officiels
with open(file_path, "r", encoding="latin-1") as file:
    column_names = file.readline().rstrip("\r\n").split("\t")

print(f"Nombre de colonnes attendues : {len(column_names)}")

df_2024 = pd.read_csv(
    file_path,
    sep="\t",
    encoding="latin-1",
    header=0,
    names=column_names,
    usecols=range(len(column_names)),
    index_col=False,
    low_memory=False,
)
```

### Observation

Les colonnes de consommation et de production ne comportent des mesures que toutes les 30 minutes.

Les créneaux intermédiaires (00:15, 00:45, 01:15, etc.) présentent des valeurs manquantes pour ces variables.

À l'inverse, les colonnes de prévision (Prévision J-1 et Prévision J) sont renseignées toutes les 15 minutes.

### Hypothèse

Cette différence semble résulter d'une règle de publication des données et non d'un décalage ou d'une erreur d'importation.
